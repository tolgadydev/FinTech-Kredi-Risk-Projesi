```python
import numpy as np
import pandas as pd

from credit_features.application import build_g1
from credit_features.behaviour import build_g2
from credit_features.bureau import build_g3

yol = "../data/parquet/"
app = pd.read_parquet(yol + "application_train.parquet")

g1 = build_g1(app)
g2 = build_g2(
    pd.read_parquet(yol + "installments_payments.parquet"),
    pd.read_parquet(yol + "POS_CASH_balance.parquet"),
    pd.read_parquet(yol + "credit_card_balance.parquet"),
    pd.read_parquet(yol + "previous_application.parquet"),
)
g3 = build_g3(
    app,
    pd.read_parquet(yol + "bureau.parquet"),
    pd.read_parquet(yol + "bureau_balance.parquet"),
)

for ad, tablo in [("G1", g1), ("G2", g2), ("G3", g3)]:
    print(ad, tablo.shape, "sonsuz değer:", int(np.isinf(tablo.select_dtypes("number")).sum().sum()))

g2.describe().T
```
